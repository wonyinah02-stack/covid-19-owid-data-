```python
#Data cleaning
```

```python
clean_df = df.copy()
```

```python
missing_values = clean_df.isna().sum()
```

```python
missing_percentage = (missing_values/len(clean_df))*100
```

```python
missing_percentage.sort_values(ascending = False)
#sorting the percentages in descending order
```

```python
filt = (missing_percentage >= 70)
columns_to_drop = missing_percentage[filt].index
#filting out the columns to drop
```

```python
# droping out columns with more than 70 percent missing values
clean_df = clean_df.drop(columns = columns_to_drop, errors = 'ignore')
```

```python
#creating new row called death rate
clean_df['death_rate'] = (clean_df['total_deaths']/clean_df['total_cases'])*1000
```

```python
clean_df.head()
```

```python
#creating new row called recovery estimate
clean_df['recovery_estimate'] = clean_df['total_cases'] - clean_df['total_deaths']
```

```python
clean_df.head()
```

```python
#creating new row called total cases per million
clean_df['total_cases_per_million'] = clean_df['total_cases']/1000000
```

```python
#creating new row called new cases per million
clean_df['new_cases_per_million'] = clean_df['new_cases']/1000000
```

```python
#creating new row called cases per million difference
clean_df['Cases_per_Million_Difference'] = clean_df['total_cases_per_million'] - clean_df['new_cases_per_million']
```

```python
#sorting
#top 20 countries with highest total cases
top_total_cases = clean_df.sort_values(by = 'total_cases', ascending = False).drop_duplicates(subset = 'country').head(20)
top_total_cases[['country', 'total_cases']]
```
